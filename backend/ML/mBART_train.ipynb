```python
import pandas as pd
import transformers
import re
```

```python
df = pd.read_excel('english-nepali-roman.xlsx')
df
```

```python
def pre_process(text):
  text = text.lower()
  text = re.sub(r'[^\w\s]', '', text) 
  text = re.sub(r'\s+', ' ', text).strip()  
  return text
```

```python
df['roman_sent']= df['roman_sent'].astype(str)
df['english_sent'] = df['english_sent'].astype(str)
```

```python
df['roman_sent'] = df['roman_sent'].apply(pre_process)
```

```python
df['english_sent'] = df['english_sent'].apply(pre_process)
```

```python
df_tmp = df[['roman_sent', 'english_sent']].copy()
df_tmp
```

```python
import pandas as pd
from datasets import Dataset
from transformers import MBart50TokenizerFast
```

```python
tokenizer = MBart50TokenizerFast.from_pretrained("facebook/mbart-large-50-many-to-many-mmt")
tokenizer.src_lang = "ne_XX"
tokenizer.tgt_lang = "en_XX"

df["roman_sent"] = ">>ne<<" + df_tmp["roman_sent"]

raw_dataset = Dataset.from_pandas(df_tmp)

def tokenize_fn(batch):
    model_inputs = tokenizer(batch["roman_sent"], max_length=128, truncation=True, padding="max_length")
    with tokenizer.as_target_tokenizer():
        labels = tokenizer(batch["english_sent"], max_length=128, truncation=True, padding="max_length")
    model_inputs["labels"] = labels["input_ids"]
    return model_inputs

tokenized_dataset = raw_dataset.map(tokenize_fn, batched=True)
```

```python
raw_dataset[2002]
```

```python
from transformers import MBartForConditionalGeneration, Seq2SeqTrainer, Seq2SeqTrainingArguments
```

```python


model = MBartForConditionalGeneration.from_pretrained("facebook/mbart-large-50-many-to-many-mmt")

training_args = Seq2SeqTrainingArguments(
    output_dir="./ronp_en_mbart",
    per_device_train_batch_size=4,
    per_device_eval_batch_size=4,
    learning_rate=3e-5,
    num_train_epochs=5,
    save_total_limit=2,
    # evaluation_strategy="epoch",
    predict_with_generate=True,
    logging_dir="./logs",
)

trainer = Seq2SeqTrainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset,
    tokenizer=tokenizer,
)

trainer.train()
```

```python
from transformers import MBartForConditionalGeneration, MBart50TokenizerFast

tokenizer = MBart50TokenizerFast.from_pretrained("facebook/mbart-large-50-many-to-many-mmt")

model = MBartForConditionalGeneration.from_pretrained("./drive/MyDrive/Checkpoints/checkpoint-17000")

tokenizer.src_lang = "ne_XX"
tokenizer.tgt_lang = "en_XX"
```

```python
from indic_transliteration.sanscript import transliterate, ITRANS, DEVANAGARI

text = transliterate('maa birami chu',  DEVANAGARI,  ITRANS)
text
```

```python
text = ">>ne<< " + text
inputs = tokenizer(pre_process(text), return_tensors="pt")
generated = model.generate(
    **inputs,
    forced_bos_token_id=tokenizer.lang_code_to_id["en_XX"],
    max_length=128,
    num_beams=5
)

print(tokenizer.decode(generated[0], skip_special_tokens=True))
```
